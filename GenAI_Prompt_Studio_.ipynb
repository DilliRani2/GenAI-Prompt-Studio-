```python
# ============================================================
# PROMPT LIBRARY
# ============================================================

class PromptLibrary:

    @staticmethod
    def summarize(text, length="short"):
        return f"""Summarize the following text in a {length} paragraph.

Text:
{text}"""

    @staticmethod
    def translate(text, target_lang):
        return f"""Translate the following text into {target_lang}.

Text:
{text}"""

    @staticmethod
    def code_explain(code):
        return f"""Explain the following code line by line.
Use simple language and include an example.

Code:
{code}"""

    @staticmethod
    def sentiment(text):
        return f"""Analyze the sentiment of the following text.
Respond only in JSON format:
{{
  "sentiment": "positive/negative/neutral",
  "confidence": "high/medium/low",
  "reason": "brief explanation"
}}

Text: {text}"""

    @staticmethod
    def quiz_question(topic, difficulty="medium"):
        return f"""Create one {difficulty}-difficulty multiple-choice
quiz question about {topic}.

Use this format:
Question: [question]
A) [option]
B) [option]
C) [option]
D) [option]
Answer: [letter]"""


# Test the existing prompt library
pl = PromptLibrary()

print("\n--- Summarization ---")
print(ask(pl.summarize(
    "Artificial intelligence enables computers to perform tasks "
    "that normally require human intelligence, including learning, "
    "reasoning, and understanding language.",
    "short"
)))

print("\n--- Sentiment Analysis ---")
print(ask(pl.sentiment(
    "The product is excellent and easy to use."
)))


# ============================================================
# PROMPT TEMPLATE CLASS
# ============================================================

class PromptTemplate:

    def __init__(self, template):
        self.template = template

    def format(self, **kwargs):
        return self.template.format(**kwargs)

    def run(self, **kwargs):
        prompt = self.format(**kwargs)
        return ask(prompt)


# ============================================================
# EXERCISE 1: CREATE 5 MORE PROMPT TEMPLATES
# ============================================================

# 1. Blog Post Template
blog_post_template = PromptTemplate("""
You are a professional blog writer.

Write a blog post about {topic}.
Target audience: {audience}
Tone: {tone}
Length: {length}

Include an engaging title, introduction, headings,
useful information, and a conclusion.
""")


# 2. Social Media Template
social_media_template = PromptTemplate("""
You are a social media content creator.

Create a social media post about {topic}.
Platform: {platform}
Tone: {tone}

Include an engaging opening, a concise caption,
a call to action, and relevant hashtags.
""")


# 3. Product Description Template
product_description_template = PromptTemplate("""
You are an experienced product copywriter.

Write a product description for {product}.
Target audience: {audience}
Key features: {features}
Tone: {tone}

Explain the product's benefits clearly.
Do not invent specifications or features
that were not provided.
""")


# 4. Interview Question Template
interview_question_template = PromptTemplate("""
You are an experienced technical interviewer.

Create {number} interview questions about {topic}.
Difficulty: {difficulty}
Role: {role}

Include a concise answer and a brief explanation
for each question.
""")


# 5. Feedback Template
feedback_template = PromptTemplate("""
You are a constructive and supportive reviewer.

Review the following work:
{work}

Context: {context}
Tone: {tone}

Provide:
1. Strengths
2. Areas for improvement
3. Specific suggestions
4. An encouraging concluding comment

Keep the feedback clear, respectful, and actionable.
""")


# ============================================================
# TEST ALL 5 NEW TEMPLATES
# ============================================================

print("\n--- 1. Blog Post ---")
print(blog_post_template.run(
    topic="Benefits of Artificial Intelligence",
    audience="college students",
    tone="friendly and informative",
    length="approximately 200 words"
))

print("\n--- 2. Social Media Post ---")
print(social_media_template.run(
    topic="Learning Python programming",
    platform="LinkedIn",
    tone="professional and encouraging"
))

print("\n--- 3. Product Description ---")
print(product_description_template.run(
    product="Wireless Bluetooth Headphones",
    audience="college students",
    features="wireless connectivity, rechargeable battery, "
             "lightweight design",
    tone="clear and persuasive"
))

print("\n--- 4. Interview Questions ---")
print(interview_question_template.run(
    number=5,
    topic="Core Java",
    difficulty="beginner",
    role="Java Developer Intern"
))

print("\n--- 5. Feedback ---")
print(feedback_template.run(
    work="A student created a simple Python calculator.",
    context="beginner programming project",
    tone="friendly and constructive"
))


# ============================================================
# EXISTING EMAIL TEMPLATE EXAMPLE
# ============================================================

email_template = PromptTemplate("""
Write a {tone} email about {topic}.
- Keep it {length}
- Include a clear call to action.
""")

print("\n--- Email Template ---")
print(email_template.run(
    tone="professional",
    topic="requesting a project deadline extension",
    length="brief, around 3-4 sentences"
))

print("\nAll prompt template exercises completed!")
```
